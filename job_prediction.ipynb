```python
import pandas as pd
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score
X = pd.get_dummies(
    df.drop(columns=["status", "placed", "academic_avg", "interview_score", "experience_category",
                     "academic_band", "skills_match_level", "interview_category"]),
    drop_first=True)
model = HistGradientBoostingClassifier(random_state=42)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
prob = cross_val_predict(model, X, df["placed"], cv=cv, method="predict_proba")[:, 1]
df["placement_probability_score"] = (prob * 100).round(1)
df["probability_band"] = pd.cut(df["placement_probability_score"], bins=[-0.1, 20, 40, 100.1],
                                labels=["Low chance", "Medium chance", "High chance"])
 
print("Out-of-fold ROC-AUC:", round(roc_auc_score(df["placed"], prob), 3), "\n")
```

```python
y = (df["status"] == "Placed").astype(int)      # Placed = 1, Not Placed = 0

# ---------- Features: ONLY the 25 original columns ----------
feature_cols = [
    "age_years", "gender", "ssc_percentage", "hsc_percentage", "degree_percentage",
    "degree_specialization", "technical_score", "aptitude_score", "communication_score",
    "skills_match_percentage", "certifications_count", "internship_experience",
    "years_of_experience", "career_switch_willingness", "relevant_experience",
    "previous_ctc_lpa", "expected_ctc_lpa", "company_tier", "job_role_match",
    "competition_level", "bond_requirement", "notice_period_days", "layoff_history",
    "employment_gap_months", "relocation_willingness",
]

X = pd.get_dummies(df[feature_cols], drop_first=True)
# ---- 3. Train/test split (80/20, keeps the 70/30 class ratio)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, stratify=y, random_state=42)
 
# ---- 4. Model
# Scaling matters for logistic regression
model = make_pipeline(StandardScaler(),
                      LogisticRegression(max_iter=1000))
 
# ---- 5. Train
model.fit(X_train, y_train)
 
# ---- 6. Evaluate
prob = model.predict_proba(X_test)[:, 1]
pred = (prob >= 0.5).astype(int)
 
print("MODEL: Logistic Regression")
print("Accuracy :", round(accuracy_score(y_test, pred), 3))
print("Precision:", round(precision_score(y_test, pred), 3))
print("Recall   :", round(recall_score(y_test, pred), 3))
print("F1 score :", round(f1_score(y_test, pred), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, prob), 3))
print("Confusion matrix (rows = actual 0/1, cols = predicted 0/1):")
print(confusion_matrix(y_test, pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix)
y = (df["status"] == "Placed").astype(int)      # Placed = 1, Not Placed = 0

# ---------- Features: ONLY the 25 original columns ----------
feature_cols = [
    "age_years", "gender", "ssc_percentage", "hsc_percentage", "degree_percentage",
    "degree_specialization", "technical_score", "aptitude_score", "communication_score",
    "skills_match_percentage", "certifications_count", "internship_experience",
    "years_of_experience", "career_switch_willingness", "relevant_experience",
    "previous_ctc_lpa", "expected_ctc_lpa", "company_tier", "job_role_match",
    "competition_level", "bond_requirement", "notice_period_days", "layoff_history",
    "employment_gap_months", "relocation_willingness",
]

X = pd.get_dummies(df[feature_cols], drop_first=True)

# ---- 3. Train/test split (80/20, keeps the 70/30 class ratio)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, stratify=y, random_state=42)

# ---- 4. Model
model = DecisionTreeClassifier(max_depth=8, class_weight="balanced", random_state=42)

# ---- 5. Train
model.fit(X_train, y_train)

# ---- 6. Evaluate
prob = model.predict_proba(X_test)[:, 1]
pred = (prob >= 0.5).astype(int)

print("MODEL: Decision Tree")
print("Accuracy :", round(accuracy_score(y_test, pred), 3))
print("Precision:", round(precision_score(y_test, pred), 3))
print("Recall   :", round(recall_score(y_test, pred), 3))
print("F1 score :", round(f1_score(y_test, pred), 3))
print("ROC-AUC  :", round(roc_auc_score(y_test, prob), 3))
print("Confusion matrix (rows = actual 0/1, cols = predicted 0/1):")
print(confusion_matrix(y_test, pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
y = (df["status"] == "Placed").astype(int)      # Placed = 1, Not Placed = 0

# ---------- Features: ONLY the 25 original columns ----------
feature_cols = [
    "age_years", "gender", "ssc_percentage", "hsc_percentage", "degree_percentage",
    "degree_specialization", "technical_score", "aptitude_score", "communication_score",
    "skills_match_percentage", "certifications_count", "internship_experience",
    "years_of_experience", "career_switch_willingness", "relevant_experience",
    "previous_ctc_lpa", "expected_ctc_lpa", "company_tier", "job_role_match",
    "competition_level", "bond_requirement", "notice_period_days", "layoff_history",
    "employment_gap_months", "relocation_willingness",
]

X = pd.get_dummies(df[feature_cols], drop_first=True)
# 2. Split (same split for every model so the comparison is fair)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

# 3. Models
models = {
    "Gradient Boosting": HistGradientBoostingClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=300, min_samples_leaf=5,n_jobs=-1, random_state=42),
    "Logistic Regression": make_pipeline(StandardScaler(),
                                         LogisticRegression(max_iter=1000)),
    "Decision Tree": DecisionTreeClassifier(max_depth=8, random_state=42)
}

# 4. Train each model and calculate the five metrics on the TEST set
rows = []
for name, model in models.items():
    model.fit(X_train, y_train)
    prob = model.predict_proba(X_test)[:, 1]   # probability of "Job Accepted"
    pred = model.predict(X_test)               # class prediction

    rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, prob),   # uses probabilities, not 0/1
    })

# 5. Build the table, best model first
results = pd.DataFrame(rows).round(3).sort_values("ROC-AUC", ascending=False)
print(results)
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import HistGradientBoostingClassifier

# 1. Load the original dataset (change the path if the file is in another folder)
df = pd.read_csv("HR_Job_Placement_cleaned.csv")
for c in df.select_dtypes(exclude="number").columns:
    df[c] = df[c].str.strip().str.title()
df = df.drop_duplicates()

# 2. Target: Placed = 1 (accept), Not Placed = 0 (reject)
y = (df["status"] == "Placed").astype(int)

# 3. Features: the 25 original columns only
FEATURES = [
    "age_years", "gender", "ssc_percentage", "hsc_percentage", "degree_percentage",
    "degree_specialization", "technical_score", "aptitude_score", "communication_score",
    "skills_match_percentage", "certifications_count", "internship_experience",
    "years_of_experience", "career_switch_willingness", "relevant_experience",
    "previous_ctc_lpa", "expected_ctc_lpa", "company_tier", "job_role_match",
    "competition_level", "bond_requirement", "notice_period_days", "layoff_history",
    "employment_gap_months", "relocation_willingness",
]
X = df[FEATURES]
num_cols = X.select_dtypes("number").columns.tolist()
cat_cols = [c for c in FEATURES if c not in num_cols]

# 4. Gradient Boosting pipeline (fills missing values and encodes text automatically)
prep = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="constant", fill_value="Unknown")),
                      ("ohe", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
])
model = Pipeline([("prep", prep),
                  ("gb", HistGradientBoostingClassifier(class_weight="balanced", random_state=42))])
model.fit(X, y)
print("Gradient Boosting model trained on", len(X), "candidates")
```

```python
import pandas as pd
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import HistGradientBoostingClassifier

# ---- Load the ORIGINAL dataset ----
df = pd.read_csv("HR_Job_Placement_cleaned.csv")
for col in df.select_dtypes(exclude="number").columns:
    df[col] = df[col].str.strip().str.title()
y = (df["status"] == "Placed").astype(int)            # Placed = accept (1), Not Placed = reject (0)

# ---- Use ONLY the 25 original feature columns (nothing else can sneak in) ----
FEATURES = [
    "age_years", "gender", "ssc_percentage", "hsc_percentage", "degree_percentage",
    "degree_specialization", "technical_score", "aptitude_score", "communication_score",
    "skills_match_percentage", "certifications_count", "internship_experience",
    "years_of_experience", "career_switch_willingness", "relevant_experience",
    "previous_ctc_lpa", "expected_ctc_lpa", "company_tier", "job_role_match",
    "competition_level", "bond_requirement", "notice_period_days", "layoff_history",
    "employment_gap_months", "relocation_willingness",
]
X = df[FEATURES]
text_cols = X.select_dtypes(exclude="number").columns.tolist()
print("Features used:", X.shape[1], "(should be 25)")

model = make_pipeline(
    make_column_transformer((OneHotEncoder(handle_unknown="ignore"), text_cols), remainder="passthrough"),
    HistGradientBoostingClassifier(class_weight="balanced", random_state=42))
model.fit(X, y)

# ---- Get input from the user ----
c = X.median(numeric_only=True).to_dict()              # typical values for the details we don't ask
c.update(X[text_cols].mode().iloc[0].to_dict())
c["technical_score"] = float(input("Technical score (0-100): "))
c["skills_match_percentage"] = float(input("Skills match % (0-100): "))
c["years_of_experience"] = int(input("Years of experience (0-5): "))
c["expected_ctc_lpa"] = float(input("Expected salary in LPA: "))
c["internship_experience"] = input("Internship done? (Yes/No): ").strip().title()
c["job_role_match"] = input("Job role matches? (Matched/Not Matched): ").strip().title()
c["competition_level"] = input("Competition level (Low/Medium/High): ").strip().title()

# ---- Predict ----
prob = model.predict_proba(pd.DataFrame([c])[FEATURES])[0, 1]
print("\nThe candidate WILL ACCEPT the job" if prob >= 0.5 else "\nThe candidate will NOT accept the job")
print(f"Model score: {prob*100:.1f}%")
```
